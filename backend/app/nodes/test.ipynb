```python
from jinja2 import Template
from typing import List, Dict, Any, Set
```

```python


template = ["{{name}}", "{{age}}"]
render_context = [
    {"name": "john", "age": 25},
    {"name": "Jane", "age": 18}
]

def render_template_sets(template: List[str], render_context: List[Dict[str, Any]]) -> List[Set[Any]]:
    result = []
    for context in render_context:
        row_set: Set[Any] = set()
        for tmpl_str in template:
            t = Template(tmpl_str)
            resolved = t.render(**context)
            # Convert to original type when possible (e.g. number)
            try:
                resolved = int(resolved) if resolved.isdigit() else resolved
            except:
                pass
            row_set.add(resolved)
        result.append(row_set)
    return result


# Example
template = ["{{name}}", "{{age}}"]
render_context = [{"name": "john", "age": 25}, {"name": "Jane", "age": 18}]

output = render_template_sets(template, render_context)
print(output)
```
